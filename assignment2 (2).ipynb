```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('simple_login_activity.csv')

print("First 5 rows:")
display(df.head())
```

```python
print("Dataset Information:")
print(df.info())

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nNumber of rows:", len(df))
print("Number of columns:", len(df.columns))
```

```python
# Remove duplicate rows
df = df.drop_duplicates().copy()

# Fill missing numeric values with median
df['failed_logins'] = df['failed_logins'].fillna(
    df['failed_logins'].median()
)

df['download_mb'] = df['download_mb'].fillna(
    df['download_mb'].median()
)

print("Missing values after cleaning:")
print(df.isna().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nFinal number of rows:", len(df))
```

```python
print("Cleaned Dataset - First 5 Rows:")
display(df.head())
```

```python
features = [
    'failed_logins',
    'login_hour',
    'new_device',
    'download_mb'
]

X = df[features]
y = df['label']

print("Features:")
print(features)

print("\nTarget:")
print("label")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Actual values:")
print(y_test.values)

print("\nPredicted values:")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=['Normal', 'Suspicious']
)

plt.title('Decision Tree Confusion Matrix')
plt.show()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Data standardized successfully!")
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

df['cluster'] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")
```

```python
cluster_summary = df.groupby('cluster')[features].mean().round(2)

print("Cluster Mean Summary:")
display(cluster_summary)
```

```python
label_summary = pd.crosstab(
    df['cluster'],
    df['label']
)

print("Cluster vs Label:")
display(label_summary)
```

```python
print("Cluster Summary:")
display(cluster_summary)

print("\nInterpretation:")
print("A cluster with higher failed logins, new-device usage, unusual login hours,")
print("or larger downloads may appear more suspicious.")
```

```python
importance = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values(
    'importance',
    ascending=False
)

print("Feature Importance:")
display(importance)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    importance['feature'],
    importance['importance']
)

plt.title('Decision Tree Feature Importance')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```
